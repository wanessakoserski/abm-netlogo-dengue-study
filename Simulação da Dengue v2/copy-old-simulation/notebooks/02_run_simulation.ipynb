```python
# Simulação Completa de Dengue

Este notebook executa simulações completas de dengue com parâmetros realistas e análise estatística.
```

```python
# Importar dependências
import sys
import os
sys.path.append('..')

from model.model import DengueABM
from visualization.plot_results import plot_model_results, plot_epidemic_curves
from analysis.evaluate_model import run_multiple_simulations, plot_results_with_ci
import time
import pandas as pd
```

```python
# Simulação única com parâmetros realistas
print("=== SIMULAÇÃO ÚNICA REALISTA ===")

model = DengueABM(
    width=50, 
    height=50,
    initial_humans=1000,
    initial_mosquitoes=5000,
    initial_infected_humans=10,
    initial_infected_mosquitoes=50,
    mosquitoes_eggs=50,
    inherit_dengue_eggs_percentage=5,
    fatality_percentage=1
)

print(f"Início: {model.people_count_with_dengue} pessoas infectadas, {model.mosquitoes_count_with_dengue} mosquitos infectados")

start_time = time.time()

# Executar por 180 dias (6 meses)
for i in range(180):
    if not model.running:
        print(f"Simulação parou no dia {i}")
        break
    model.step()
    if i % 30 == 0:  # Relatório mensal
        print(f"Mês {i//30 + 1}: {model.people_count_with_dengue} pessoas infectadas, {model.mosquitoes_count_with_dengue} mosquitos infectados")

end_time = time.time()
print(f"\nTempo de execução: {end_time - start_time:.2f} segundos")
print(f"Resultado final: {model.people_count_with_dengue} pessoas infectadas, {model.mosquitoes_count_with_dengue} mosquitos infectados")
```

```python
# Visualizar resultados
plot_model_results(model, "results/notebook_simulation.png")
plot_epidemic_curves(model, "results/notebook_curves.png")
```

```python
# Análise estatística com múltiplas simulações
print("\n=== ANÁLISE ESTATÍSTICA ===")

params = {
    "width": 50,
    "height": 50,
    "initial_humans": 1000,
    "initial_mosquitoes": 5000,
    "initial_infected_humans": 10,
    "initial_infected_mosquitoes": 50,
    "mosquitoes_eggs": 50,
    "inherit_dengue_eggs_percentage": 5,
    "fatality_percentage": 1
}

print("Executando 10 simulações para análise estatística...")
results_df = run_multiple_simulations(DengueABM, params, n_simulations=10, steps=180)

# Visualizar resultados estatísticos
plot_results_with_ci(results_df, "total_infected", "Número Total de Infectados")
plot_results_with_ci(results_df, "total_deaths_dengue", "Número de Mortes por Dengue")

# Salvar resultados
results_df.to_csv("results/notebook_multiple_simulations.csv", index=False)
print("Resultados salvos em results/notebook_multiple_simulations.csv")
```
